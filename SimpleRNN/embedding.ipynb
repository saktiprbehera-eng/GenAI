```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### Sentences
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good',
]
```

```python
sent
```

```python
## Define the vocabulary size
vocab_size=10000
```

```python
### One hot representation of the sentences
one_hot_repr=[one_hot(word,vocab_size) for word in sent]
one_hot_repr
```

```python
from tensorflow.keras.layers import Embedding
# from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sent_len=8
embedded_doc=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_len)
embedded_doc
```

```python
## Feature representation of the sentences
dim=10
model=Sequential()
model.add(Embedding(vocab_size,dim,input_shape=(sent_len,)))
model.compile('adam','mse')
model.summary()
```

```python
### Get the feature representation of the sentences
feature_repr=model.predict(embedded_doc)
feature_repr
```
